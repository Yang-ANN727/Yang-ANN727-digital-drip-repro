复现流程示例：读取数据 -> 筛选 -> 描述性统计 -> statsmodels 回归（含 firm/year 固定效应，按 firm 聚类标准误）


```python
import pandas as pd
import statsmodels.formula.api as smf
from pathlib import Path
print('请先确认 conda 环境已激活并选择了 kernel: digital-drip')
```

```python
# 读取数据（示例）
p = Path('data/raw_data.csv')
if p.exists():
    df = pd.read_csv(p)
else:
    print('请将数据放到 data/raw_data.csv 或修改路径')
```

```python
# 示例筛选和回归（请根据你的列名修改）
df = df[(df['year'] >= 2010) & (df['year'] <= 2018)]
df = df.dropna(subset=['outcome', 'treatment'])
formula = 'outcome ~ treatment + cov1 + cov2 + C(firm) + C(year)'
res = smf.ols(formula, data=df).fit(cov_type='cluster', cov_kwds={'groups': df['firm']})
print(res.summary())
```
